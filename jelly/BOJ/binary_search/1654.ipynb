```python
k, n = map(int, input().split())

lines = [int(input()) for _ in range(k)]

start = 1
end = max(lines)

result = 0
while start <= end:
    mid = (start + end) // 2
    total = sum((h // mid) for h in lines)

    if total >= n:
        result = mid
        start = mid + 1
    else:
        end = mid - 1

print(result)
```
