```python
n = int(input())
budgets = list(map(int, input().split()))
total_budget = int(input())

start = 0
end = max(budgets)

result = 0
while start <= end:
    mid = (start + end) // 2

    total = 0
    for budget in budgets:
        if budget > mid:
            total += mid
        else:
            total += budget

    if total <= total_budget:
        result = mid
        start = mid + 1
    else:
        end = mid - 1

print(result)
```

```python
n = int(input())
budgets = list(map(int, input().split()))
total_budget = int(input())

start = 0
end = max(budgets)

result = 0
while start <= end:
    mid = (start + end) // 2
    total = sum(min(b, mid) for b in budgets)

    if total <= total_budget:
        result = mid
        start = mid + 1
    else:
        end = mid - 1

print(result)
```
