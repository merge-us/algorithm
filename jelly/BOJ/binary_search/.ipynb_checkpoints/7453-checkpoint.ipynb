```python
from collections import defaultdict

n = int(input())

a = []
b = []
c = []
d = []

for _ in range(n):
    row = list(map(int, input().split()))
    a.append(row[0])
    b.append(row[1])
    c.append(row[2])
    d.append(row[3])

def get_sum_count(arr_1, arr_2):
    sum_count = defaultdict(int)

    for i in arr_1:
        for j in arr_2:
            sum_count[i + j] += 1

    return sum_count

ab_sum_count = get_sum_count(a, b)
cd_sum_count = get_sum_count(c, d)

result = 0
for num, cnt in ab_sum_count.items():
    result += cnt * cd_sum_count.get(-1 * num, 0)

print(result)
```

```python
from collections import Counter
import sys

input = sys.stdin.readline

n = int(input())

A = []
B = []
C = []
D = []

for _ in range(n):
    row = list(map(int, input().split()))
    A.append(row[0])
    B.append(row[1])
    C.append(row[2])
    D.append(row[3])

ab_sum_count = Counter(a + b for a in A for b in B)
cd_sum_count = Counter(c + d for c in C for d in D)

result = 0
for num, cnt in ab_sum_count.items():
    result += cnt * cd_sum_count.get(-1 * num, 0)

print(result)
```

```python
from collections import Counter
import sys

input = sys.stdin.readline

n = int(input())

A, B, C, D = [], [], [], []

for _ in range(n):
    a, b, c, d = map(int, input().split())
    A.append(a)
    B.append(b)
    C.append(c)
    D.append(d)

ab_sum_count = Counter()
for a in A:
    for b in B:
        ab_sum_count[a + b] += 1

result = 0
for c in C:
    for d in D:
        result += ab_sum_count[-(c + d)]

print(result)
```

```python
from bisect import bisect_left, bisect_right

n = int(input())

A, B, C, D = [], [], [], []

for _ in range(n):
    a, b, c, d = map(int, input().split())
    A.append(a)
    B.append(b)
    C.append(c)
    D.append(d)

AB = []
for a in A:
    for b in B:
        AB.append(a + b)

CD = []
for c in C:
    for d in D:
        CD.append(c + d)
        
AB.sort()
CD.sort()


result = 0
for ab in AB:
     result += bisect_right(CD, -ab) - bisect_left(CD, -ab)

print(result)
```

```python
from bisect import bisect_left, bisect_right
import sys

input = sys.stdin.readline

n = int(input())

A, B, C, D = [], [], [], []

for _ in range(n):
    a, b, c, d = map(int, input().split())
    A.append(a)
    B.append(b)
    C.append(c)
    D.append(d)

ab_sum = [a + b for a in A for b in B]
cd_sum = [c + d for c in C for d in D]
        
ab_sum.sort()
cd_sum.sort()

result = 0
for ab in ab_sum:
     result += bisect_right(cd_sum, -ab) - bisect_left(cd_sum, -ab)

print(result)
```
