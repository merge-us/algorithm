```python
from bisect import bisect_left

t = int(input())

for _ in range(t):
    n, m = map(int, input().split())

    a_s = list(map(int, input().split()))
    b_s = list(map(int, input().split()))
    b_s.sort()

    result = 0
    for a in a_s:
        result += bisect_left(b_s, a)

    print(result)
```
