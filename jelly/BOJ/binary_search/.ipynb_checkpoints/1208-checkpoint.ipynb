```python
from collections import Counter

def dfs(idx, current, arr, result):

    if idx == len(arr):
        result.append(current)
        return 

    dfs(idx + 1, current, arr, result)
    dfs(idx + 1, current + arr[idx], arr, result)

    
n, s = map(int, input().split())
array = list(map(int, input().split()))

arr1 = array[:n//2]
arr2 = array[n//2:]

result1 = []
dfs(0, 0, arr1, result1)
counter1 = Counter(result1)

result2 = []
dfs(0, 0, arr2, result2)
counter2 = Counter(result2)

total_count = 0
for i, cnt in counter1.items():
    total_count += cnt * counter2.get(-i, 0)

if s == 0:
    total_count -= 1

print(total_count)
```

```python
from collections import Counter

def get_sub_sums(arr):
    result = []
    
    def dfs(idx, current):
        if idx == len(arr):
            result.append(current)
            return 
    
        dfs(idx + 1, current)
        dfs(idx + 1, current + arr[idx])

    dfs(0, 0)
    return result
    
n, s = map(int, input().split())
nums = list(map(int, input().split()))

left = nums[:n//2]
right = nums[n//2:]

sum_left = get_sub_sums(left)
sum_right = get_sub_sums(right)

left_counter = Counter(sum_left)

result = 0
for r in sum_right:
    result += left_counter.get(s - r, 0)

if s == 0:
    result -= 1

print(result)
```
