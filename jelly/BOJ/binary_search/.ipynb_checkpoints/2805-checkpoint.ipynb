```python
n, m = map(int, input().split())
heights = list(map(int, input().split()))

max_height = max(heights)

start = 0
end = max_height

result = 0
while start <= end:
    mid = (start + end) // 2

    current = 0
    for height in heights:
        if height - mid > 0:
            current += height - mid

    if current >= m:
        result = mid
        start = mid + 1
    else:
        end = mid - 1

print(result)
```

```python
n, m = map(int, input().split())
heights = list(map(int, input().split()))

start = 0
end = max(heights)

result = 0
while start <= end:
    mid = (start + end) // 2
    total = sum(h - mid for h in heights if h > mid)

    if total >= m:
        result = mid
        start = mid + 1
    else:
        end = mid - 1

print(result)
```
