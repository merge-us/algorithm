```python
t = int(input())

for _ in range(t):
    n = int(input())
    memo = list(map(int, input().split()))
    memo.sort()

    m = int(input())
    targets = list(map(int, input().split()))

    for target in targets:
        start = 0
        end = n - 1

        exist = False
        while start <= end:
            mid = (start + end) // 2

            if memo[mid] == target:
                exist = True
                break
            elif memo[mid] < target:
                start = mid + 1
            else:
                end = mid - 1

        print(1 if exist else 0)
```

```python
t = int(input())

for _ in range(t):
    n = int(input())
    memo = set(map(int, input().split()))
    
    m = int(input())
    targets = list(map(int, input().split()))

    for target in targets:
         print(1 if target in memo else 0)
```
