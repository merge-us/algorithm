```python
from bisect import bisect_left
import sys

input = sys.stdin.readline

n, h = map(int, input().split())

floors = []
ceils = []

for i in range(n):
    obstacle = int(input())
    if i % 2 == 0:
        floors.append(obstacle)
    else:
        ceils.append(obstacle)

floors.sort()
ceils.sort()

min_obstacle = n
obstacle_count = 0
for i in range(1, h + 1):
    floor_count = (n // 2) - bisect_left(floors, i)
    ceil_count = (n // 2) - bisect_left(ceils, h - i + 1)

    current = floor_count + ceil_count

    if current == min_obstacle:
        obstacle_count += 1
    elif current < min_obstacle:
        min_obstacle = current
        obstacle_count = 1
    

print(min_obstacle, obstacle_count)
```

```python
from bisect import bisect_left
import sys

input = sys.stdin.readline

n, h = map(int, input().split())

floors = []
ceils = []

for i in range(n):
    obstacle = int(input())
    if i % 2 == 0:
        floors.append(obstacle)
    else:
        ceils.append(obstacle)

floors.sort()
ceils.sort()

min_obstacle = n
obstacle_count = 0

len_f = len(floors)
len_c = len(ceils)

for i in range(1, h + 1):
    floor_crash = len_f - bisect_left(floors, i)
    ceil_crash = len_c - bisect_left(ceils, h - i + 1)

    crash_count = floor_crash + ceil_crash

    if crash_count == min_obstacle:
        obstacle_count += 1
    elif crash_count < min_obstacle:
        min_obstacle = crash_count
        obstacle_count = 1
    
print(min_obstacle, obstacle_count)
```
