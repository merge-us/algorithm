```python
n = int(input())
data = list(map(int, input().split()))

total_min = float('inf')
result = (-1, n + 1)

for i in range(n - 1):
    current = data[i]

    start = i + 1
    end = n - 1

    current_min = float('inf')
    target = i
    while start <= end:
        mid = (start + end) // 2

        if abs(current + data[mid]) < current_min:
            current_min = abs(current + data[mid])
            target = mid
            
        if current + data[mid] < 0:
            start = mid + 1
        else:
            end = mid - 1

    if current_min < total_min:
        total_min = current_min
        result = (data[i], data[target])

print(result[0], end=" ")
print(result[1])
        
        
```

```python
n = int(input())
data = list(map(int, input().split()))

total_min = float('inf')
result = (0, 0)

for i in range(n - 1):
    current = data[i]

    start = i + 1
    end = n - 1

    while start <= end:
        mid = (start + end) // 2
        mix = current + data[mid]
        
        if abs(mix) < total_min:
            total_min = abs(mix)
            result = (current, data[mid])
            
        if mix < 0:
            start = mid + 1
        else:
            end = mid - 1

print(*result)
```
