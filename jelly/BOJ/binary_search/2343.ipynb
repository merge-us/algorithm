```python
n, m = map(int, input().split())
records = list(map(int, input().split()))

avg_cnt = n // m
if n % m > 0:
    avg_cnt += 1

start = avg_cnt * 1
end = avg_cnt * 10000

result = 0
while start <= end:
    target_length = (start + end) // 2
    
    count = 0
    current_length = 0
    for record in records:
        current_length += record
        if current_length > target_length:
            count += 1
            current_length = record

    if current_length == records[-1]:
        count += 1
        
    if count <= m:
        result = target_length
        end = target_length - 1
    else:
        start = target_length + 1

print(result)
        


```

```python
n, m = map(int, input().split())
records = list(map(int, input().split()))

start = max(records)
end = sum(records)

result = 0
while start <= end:
    target_length = (start + end) // 2
    
    count = 1
    current_length = 0
    for record in records:
        if current_length + record > target_length:
            count += 1
            current_length = record
        else:
            current_length += record

    if count <= m:
        result = target_length
        end = target_length - 1
    else:
        start = target_length + 1

print(result)
```
