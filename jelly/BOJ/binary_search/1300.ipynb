```python
n = int(input())
k = int(input())

def count(x):
    total = 0
    for i in range(1, n + 1):
        total += min(x // i, n)
    return total

start = 1
end = k
result = 0

while start <= end:
    mid = (start + end) // 2

    if count(mid) >= k:
        result = mid
        end = mid - 1
    else:
        start = mid + 1

print(result)

```
