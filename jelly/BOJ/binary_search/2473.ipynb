```python
n = int(input())
data = list(map(int, input().split()))

data.sort()

min_mix = float('inf')
results = (0, 0, 0)
for i in range(n):
    current = data[i]
    
    left = 0
    right = n - 1

    while left < right:
        if left == i:
            left += 1
            
        if right == i:
            right -= 1

        if left >= right:
            break

        mix = current + data[left] + data[right]

        if abs(mix) < min_mix:
            min_mix = abs(mix)
            results = sorted((current, data[left], data[right]))

        if mix < 0:
            left += 1
        else:
            right -= 1

print(*results)
```
