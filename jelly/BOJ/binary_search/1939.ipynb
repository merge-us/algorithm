```python
from collections import deque
import sys

input = sys.stdin.readline

n, m = map(int, input().split())
edges = [tuple(map(int, input().split())) for _ in range(m)]
start_node, end_node = map(int, input().split())

def is_connected(graph, start, end):
    visited = [False] * (n + 1)
    
    q = deque([])
    q.append(start)
    visited[start] = True

    while q:
        current = q.popleft()
        if current == end:
            return True

        for i in graph[current]:
            if not visited[i]:
                visited[i] = True
                q.append(i)

    return False

start = 1
end = 1_000_000_000

result = 0
while start <= end:
    mid = (start + end) // 2

    graph = [[] for _ in range(n + 1)]
    for a, b, c in edges:
        if c >= mid:
            graph[a].append(b)
            graph[b].append(a)

    if is_connected(graph, start_node, end_node):
        result = mid
        start = mid + 1
    else:
        end = mid - 1

print(result)
```
