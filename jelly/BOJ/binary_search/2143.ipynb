```python
from collections import defaultdict

t = int(input())

n = int(input())
a = list(map(int, input().split()))

m = int(input())
b = list(map(int, input().split()))


def count_sub_arr_sum(size, target):
    result = defaultdict(int)

    for i in range(size):
        current = target[i]
        result[current] += 1
        for j in range(i + 1, size):
            current += target[j]
            result[current] += 1

    return result

dict_a = count_sub_arr_sum(n, a)
dict_b = count_sub_arr_sum(m, b)

keys_b = list(dict_b.keys())
keys_b.sort()
len_b = len(keys_b)

count = 0
for key_a in dict_a.keys():
    start = 0
    end = len_b - 1

    while start <= end:
        mid = (start + end) // 2

        if keys_b[mid] == t - key_a:
            count += dict_a[key_a] * dict_b[keys_b[mid]]
            break
        elif keys_b[mid] < t - key_a:
            start = mid + 1
        else:
            end = mid - 1

print(count)
```

```python
from collections import defaultdict

t = int(input())

n = int(input())
a = list(map(int, input().split()))

m = int(input())
b = list(map(int, input().split()))


def get_sub_sum_count(arr):
    sub_sum_count = defaultdict(int)

    for i in range(len(arr)):
        current = 0
        for j in range(i, len(arr)):
            current += arr[j]
            sub_sum_count[current] += 1

    return sub_sum_count

dict_a = get_sub_sum_count(a)
dict_b = get_sub_sum_count(b)

result = 0
for a_sum, a_cnt in dict_a.items():
    b_target = t - a_sum
    result += a_cnt * dict_b.get(b_target, 0)

print(result)
```
