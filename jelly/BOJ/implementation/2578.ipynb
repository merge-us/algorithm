```python
import sys

board = [list(map(int, input().split())) for _ in range(5)]
calls = [list(map(int, input().split())) for _ in range(5)]

def find_pos(target):
    for i in range(5):
        for j in range(5):
            if board[i][j] == target:
                return (i, j)

row_bingo = [0] * 5
col_bingo = [0] * 5
cross_1 = set((i, i) for i in range(5))
cross_2 = set((i, 5 - i - 1) for i in range(5))

for i in range(5):
    for j in range(5):
        x, y = find_pos(calls[i][j])
        row_bingo[x] += 1
        col_bingo[y] += 1

        if (x, y) in cross_1:
            cross_1.remove((x, y))

        if (x, y) in cross_2:
            cross_2.remove((x, y))

        count = 0
        if len(cross_1) == 0:
            count += 1
    
        if len(cross_2) == 0:
            count += 1
        count += row_bingo.count(5) + col_bingo.count(5)
    
    
        if count >= 3:
            print(5 * i + (j + 1))
            sys.exit()
    
        
```

```python
import sys

position = {}
for i in range(5):
    row = list(map(int, input().split()))
    for j in range(5):
        position[row[j]] = (i, j)
        
calls = [list(map(int, input().split())) for _ in range(5)]

row_bingo = [0] * 5
col_bingo = [0] * 5
cross_1 = 0
cross_2 = 0

for i in range(5):
    for j in range(5):
        turn = (5 * i) + (j + 1)
        
        x, y = position[calls[i][j]]
        row_bingo[x] += 1
        col_bingo[y] += 1

        if x == y:
            cross_1 += 1
        if x + y == 4:
            cross_2 += 1

        bingo = 0
        bingo += row_bingo.count(5)
        bingo += col_bingo.count(5)
        bingo += (1 if cross_1 == 5 else 0)
        bingo += (1 if cross_2 == 5 else 0)
            
        if bingo >= 3:
            print(turn)
            sys.exit()
```
