```python
from collections import deque

n, m, k = map(int, input().split())

board = []
board.append([0] * (m + 1))
for _ in range(n):
    board.append([0] + list(map(int, input().split())))

def get_score(x, y, score):
    visited = [[False] * (m + 1) for _ in range(n + 1)]
    q = deque([])
    q.append((x, y))
    visited[x][y] = True
    result = 1

    while q:
        x, y = q.popleft()

        for dx, dy in directions:
            nx = x + dx
            ny = y + dy

            if 0 < nx <= n and 0 < ny <= m and not visited[nx][ny] and board[nx][ny] == score:
                q.append((nx, ny))
                visited[nx][ny] = True
                result += 1

    return result * score

def roll_dice(direction):
    dx, dy = directions[direction]

    q = deque([])

    if dx == 0:
        for i in range(3):
            q.append(dice[1][i])
        q.rotate(dy)

        for i in range(3):
            dice[1][i] = q[i]

        if dy == 1:
            dice[1][0], dice[3][1] = dice[3][1], dice[1][0]
        else:
            dice[1][2], dice[3][1] = dice[3][1], dice[1][2]
    else:
        for i in range(4):
            q.append(dice[i][1])
        q.rotate(dx)

        for i in range(4):
            dice[i][1] = q[i]
        
        
#동 남 서 북
directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]
direction = 0
dice = [[0, 2, 0],
        [4, 1, 3],
        [0, 5, 0],
        [0, 6, 0]]

x, y = 1, 1

total_score = 0
for _ in range(k):
    dx, dy = directions[direction]
    nx = x + dx
    ny = y + dy

    if not (0 < nx <= n and 0 < ny <= m):
        direction = (direction + 2) % 4
        dx, dy = directions[direction]
        nx = x + dx
        ny = y + dy

    x, y = nx, ny
    score = board[x][y]
    
    total_score += get_score(x, y, score)

    roll_dice(direction)
    bottom = dice[3][1]

    if bottom > score:
        direction = (direction + 1) % 4
    elif bottom < score:
        direction = (direction - 1) % 4

print(total_score)  
```
