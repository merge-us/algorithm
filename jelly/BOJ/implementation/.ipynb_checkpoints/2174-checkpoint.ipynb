```python
a, b = map(int, input().split())

board = []
board.append([-1] * (b + 1))
for _ in range(a):
    board.append([-1] + [0] * b)

n, m = map(int, input().split())
robots = [None] * (n + 1)

def get_direction(s):
    if s == 'N':
        return 0
    elif s == 'E':
        return 1
    elif s == 'S':
        return 2
    else:
        return 3
        
#북-동-남-서
directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]

for i in range(n):
    x, y, d = input().split()
    x, y = int(x), int(y)
    robots[i + 1] = (i_x, i_y, get_direction(d))
    board[i_x][i_y] = i + 1

result = ""
for _ in range(m):
    target, cmd, iters = input().split()

    if result:
        continue
    x, y, d = robots[int(target)]
    i_iters = int(iters)

    for _ in range(i_iters):
        if cmd == 'L':
            d = (d - 1) % 4
            robots[int(target)] = (x, y, d)
        elif cmd == 'R':
            d = (d + 1) % 4
            robots[int(target)] = (x, y, d)
        else:
            dx, dy = directions[d]
            nx = x + dx
            ny = y + dy
            
            if not (1 <= nx <= a and 1 <= ny <= b):
                result = "Robot " + target + " crashes into the wall"
                break
            elif board[nx][ny] > 0:
                result = "Robot " + target + " crashes into robot " + str(board[nx][ny])
                break
            else:
                robots[int(target)] = (nx, ny, d)
                board[x][y] = 0
                board[nx][ny] = int(target)
                x = nx
                y = ny
                
print(result)
```

```python
import sys

a, b = map(int, input().split())

board = [[0] * (a + 1) for _ in range(b + 1)]

n, m = map(int, input().split())
robots = [None] * (n + 1)

direction_map = {'N': 0, 'E': 1, 'S': 2, 'W': 3}
directions = [(-1, 0), (0, 1), (1, 0), (0, -1)]

for i in range(n):
    x, y, d = input().split()
    row, col = b - int(y) + 1, int(x)
     
    robots[i + 1] = (row, col, direction_map[d])
    board[row][col] = i + 1


for _ in range(m):
    robot_id, cmd, repeat = input().split()
    robot_id = int(robot_id)
    repeat = int(repeat)
    
    x, y, d = robots[robot_id]
    for _ in range(repeat):
        if cmd == 'L':
            d = (d - 1) % 4
        elif cmd == 'R':
            d = (d + 1) % 4
        elif cmd == 'F':
            dx, dy = directions[d]
            nx = x + dx
            ny = y + dy

            if not (0 < nx <= b and 0 < ny <= a):
                print(f"Robot {robot_id} crashes into the wall")
                sys.exit()
            if board[nx][ny] > 0:
                print(f"Robot {robot_id} crashes into robot {board[nx][ny]}")
                sys.exit()
            
            board[x][y] = 0
            board[nx][ny] = robot_id
            x, y = nx, ny

    robots[robot_id] = (x, y, d)

print("OK")
```
