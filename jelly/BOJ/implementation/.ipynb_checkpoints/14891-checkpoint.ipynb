```python
from collections import deque

wheels = []

for _ in range(4):
    wheels.append(deque(map(int, input())))
k = int(input())

for _ in range(k):
    wheel, direction = map(int, input().split())
    wheel -= 1

    visited = [False] * 4
    q = deque([])
    q.append((wheel, direction))
    
    while q:
        current_w, current_d = q.popleft()
        visited[current_w] = True
        
        if current_w + 1 < 4 and not visited[current_w + 1]:
            if wheels[current_w][2] + wheels[current_w + 1][6] == 1:
                q.append((current_w + 1, - current_d))
    
        if 0 <= current_w - 1 and not visited[current_w - 1]:
            if wheels[current_w][6] + wheels[current_w - 1][2] == 1:
                q.append((current_w - 1, - current_d))

        wheels[current_w].rotate(current_d)

score = 0
for i in range(4):
    score += wheels[i][0] * (2 ** i)

print(score)
    

    
    
```

```python
from collections import deque

wheels = []

for _ in range(4):
    wheels.append(deque(map(int, input())))
k = int(input())

for _ in range(k):
    wheel, direction = map(int, input().split())
    wheel -= 1

    rotate = [0] * 4
    rotate[wheel] = direction
    
    for i in range(wheel, 3):
        if wheels[i][2] != wheels[i + 1][6]:
            rotate[i + 1] = - rotate[i]
        else:
            break

    for i in range(wheel, 0, -1):
        if wheels[i][6] != wheels[i - 1][2]:
            rotate[i - 1] = - rotate[i]
        else:
            break

    for i in range(4):
        wheels[i].rotate(rotate[i])

score = 0
for i in range(4):
    score += wheels[i][0] * (2 ** i)

print(score)  
```

```python
from collections import deque

wheels = []

for _ in range(4):
    wheels.append(deque(map(int, input())))
k = int(input())

for _ in range(k):
    wheel, direction = map(int, input().split())
    wheel -= 1

    rotations = [0] * 4
    rotations[wheel] = direction

    for i in range(wheel - 1, -1, -1):
        if wheels[i][2] != wheels[i + 1][6]:
            rotations[i] = -rotations[i + 1]
        else:
            break
    
    for i in range(wheel + 1, 4):
        if wheels[i][6] != wheels[i - 1][2]:
            rotations[i] = -rotations[i - 1]
        else:
            break

    for i in range(4):
        wheels[i].rotate(rotations[i])

score = 0
for i in range(4):
    score += wheels[i][0] * (2 ** i)

print(score)  
```
