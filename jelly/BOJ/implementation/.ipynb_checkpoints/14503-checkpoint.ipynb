```python
n, m = map(int, input().split())
x, y, direction = map(int, input().split())
board = [list(map(int, input().split())) for _ in range(n)]

directions = [(-1, 0), (0, 1), (1, 0), (0, -1)]
result = 0
while True:
    if board[x][y] == 0:
        board[x][y] = -1
        result += 1

    exist = False
    for i in range(4):
        dx, dy = directions[i]
        nx = x + dx
        ny = y + dy

        if 0 <= nx < n and 0 <= ny < m and board[nx][ny] == 0:
            exist = True
            break

    if exist:
        for _ in range(4):
            direction = (direction - 1) % 4
            dx, dy = directions[direction]
            nx = x + dx
            ny = y + dy

            if 0 <= nx < n and 0 <= ny < m and board[nx][ny] == 0:
                x = nx
                y = ny
                break
    else:
        dx, dy = directions[direction]
        nx = x - dx
        ny = y - dy

        if not (0 <= nx < n and 0 <= ny < m) or board[nx][ny] == 1:
            break

        x = nx
        y = ny

print(result)
        
```

```python
n, m = map(int, input().split())
x, y, d = map(int, input().split())
board = [list(map(int, input().split())) for _ in range(n)]

directions = [(-1, 0), (0, 1), (1, 0), (0, -1)]

result = 0
while True:
    if board[x][y] == 0:
        board[x][y] = -1
        result += 1

    for _ in range(4):
        d = (d - 1) % 4
        dx, dy = directions[d]
        nx, ny = x + dx, y + dy
        
        if 0 <= nx < n and 0 <= ny < m and board[nx][ny] == 0:
            x, y = nx, ny
            break
    else:
        dx, dy = directions[d]
        nx, ny = x - dx, y - dy

        if not (0 <= nx < n and 0 <= ny < m) or board[nx][ny] == 1:
            break

        x, y = nx, ny

print(result)
        
```
