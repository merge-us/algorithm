```python
n, k = map(int, input().split())

coins = []
for _ in range(n):
    coins.append(int(input()))

result = 0
for coin in sorted(coins, reverse=True):
    result += (k // coin)
    k %= coin

print(result)
```

```python
n, k = map(int, input().split())
coins = [int(input()) for _ in range(n)]

count = 0
for coin in coins[::-1]:
    if k == 0:
        break
        
    count += (k // coin)
    k %= coin

print(count)
```
