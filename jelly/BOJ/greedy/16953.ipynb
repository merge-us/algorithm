```python
start, end = map(int, input().split())

count = 0
while start < end:
    first = end % 10
    count += 1
    if first == 1:
        end = end // 10
    elif first % 2 == 0:
        end //= 2
    else:
        break
        
if start != end:
    print(-1)
else:
    print(count + 1)
```

```python
a, b = map(int, input().split())

count = 0
while b > a:
    count += 1
    if b % 10 == 1:
        b //= 10
    elif b % 2 == 0:
        b //= 2
    else:
        break

print(count + 1 if a == b else -1)        
```
