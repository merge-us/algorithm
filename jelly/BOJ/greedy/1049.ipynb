```python
n, m = map(int, input().split())

packages = []
pieces = []
for _ in range(m):
    package, piece = map(int, input().split())
    packages.append(package)
    pieces.append(piece)

packages_min = min(packages)
pieces_min = min(pieces)

case1 = ((n // 6) + 1) * packages_min
case2 = (n // 6) * packages_min + (n % 6) * pieces_min
case3 = n * pieces_min

print(min(case1, case2, case3))
```
