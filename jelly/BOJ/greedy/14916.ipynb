```python
n = int(input())

five = n // 5
two = (n - five * 5) // 2

found = False
while five >= 0:
    if five * 5 + two * 2 == n:
        found = True
        break
    else:
        five -= 1
        two += 1

print(five + two if found else -1)
        
```

```python
n = int(input())

five = n // 5

while five >= 0:
    remain = n - (five * 5)
    if remain % 2 == 0:
        two = remain // 2
        print(five + two)
        break
    five -= 1
else:
    print(-1)
        
```
