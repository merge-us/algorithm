```python
n = int(input())

gt_one = []
one = []
lte_zero = []

for _ in range(n):
    num = int(input())
    if num > 1:
        gt_one.append(num)
    elif num == 1:
        one.append(1)
    else:
        lte_zero.append(num)

total = 0
def calc_max(array):
    global total
    
    for i in range(0, len(array), 2):
        if i == len(array) - 1:
            total += array[i]
        else:
            total += array[i] * array[i + 1]
            
calc_max(sorted(gt_one, reverse=True))
calc_max(sorted(lte_zero))
total += len(one)

print(total)
```

```python
n = int(input())

positives = []
ones = []
non_positives = []

for _ in range(n):
    num = int(input())
    if num > 1:
        positives.append(num)
    elif num == 1:
        ones.append(1)
    else:
        non_positives.append(num)

def calc_max(array):
    result = 0
    
    for i in range(0, len(array), 2):
        if i + 1 < len(array):
            result += array[i] * array[i + 1]
        else:
            result += array[i]
    
    return result
    
total = 0
total += calc_max(sorted(positives, reverse=True))
total += calc_max(sorted(non_positives))
total += sum(ones)

print(total)
```
