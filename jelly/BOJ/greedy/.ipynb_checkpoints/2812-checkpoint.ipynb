```python
n, k = map(int, input().split())

data = input()

index = 0
max_index = 0

current_max = "0"
result = ""
while index < n and k > 0:
    for _ in range(k):
        if data[index] > current_max:
            max_index = index
            current_max = data[index]
            index += 1

    result += data[max_index]
    k -= max_index

    max_index = index
    current_max = "0"

print(result)

```

```python
n, k = map(int, input().split())

data = input()

index = 0

result = ""
while k > 0:
    max_value = data[index]
    max_index = index
    
    for i in range(index, index + k + 1):
        if data[i] > max_value:
            max_value = data[i]
            max_index = i

    result += max_value
    k -= (max_index - index)
    index = max_index + 1

for i in range(index, n):
    result += data[i]
    
print(result)

```

```python
n, k = map(int, input().split())

data = list(input())
data.reverse()

result = [data.pop()]

while data:
    current = data.pop()

    while k > 0 and result and result[-1] < current:
        result.pop()
        k -= 1

    result.append(current)

if k > 0:
    result = result[:-k]

print("".join(result))    
```

```python
n, k = map(int, input().split())
number = input()

stack = []

for digit in number:
    while k > 0 and stack and stack[-1] < digit:
        stack.pop()
        k -= 1

    stack.append(digit)

if k > 0:
    stack = stack[:-k]

print("".join(stack))    
```
