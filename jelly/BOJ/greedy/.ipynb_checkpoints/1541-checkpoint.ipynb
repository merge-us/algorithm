```python
data = input()

preprocessed = ""

c_operand = ""
c_operator = ""

for c in data:
    if c is not in ["+", "-"]:
        c_operand += c
    else:
        proprocessed += int(c_operand)
        c_operand = ""
        if c == "-":
            c_operator = "-"    
        else:
            if c_operator == "-":
                c_operator = "-"
            else:
                
                c_operator = "+"
        preprocessed += c_operator            
```
