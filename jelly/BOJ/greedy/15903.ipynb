```python
import heapq

n, m = map(int, input().split())

data = list(map(int, input().split()))
data.sort()

for _ in range(m):
    first = heapq.heappop(data)
    second = heapq.heappop(data)

    heapq.heappush(data, first + second)
    heapq.heappush(data, first + second)

print(sum(data))
```

```python
import heapq

n, m = map(int, input().split())

data = list(map(int, input().split()))
heapq.heapify(data)

for _ in range(m):
    first = heapq.heappop(data)
    second = heapq.heappop(data)
    merged = first + second
    
    heapq.heappush(data, merged)
    heapq.heappush(data, merged)

print(sum(data))
```
