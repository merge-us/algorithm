```python
import heapq

n = int(input())

classes = [tuple(map(int, input().split())) for _ in range(n)]
classes.sort(key=lambda x:x[1])

rooms = []
rooms.append(0)

for s, t in classes:
    min_t = rooms[0]

    if s < min_t:
        heapq.heappush(rooms, t)
    else:
        heapq.heappop(rooms)
        heapq.heappush(rooms, min_t + t)

    print(rooms)

print(len(rooms))
```

```python
import heapq
import sys

input = sys.stdin.readline

n = int(input())

classes = [tuple(map(int, input().split())) for _ in range(n)]
classes.sort()

rooms = []

for s, t in classes:
    if rooms and rooms[0] <= s:
        heapq.heappop(rooms)
    heapq.heappush(rooms, t)

print(len(rooms))
```
