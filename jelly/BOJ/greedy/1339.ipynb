```python
n = int(input())

data = {}
for _ in range(n):
    num = input()

    digits = len(num)
    for i in range(digits):
        if num[i] not in data:
            data[num[i]] = 10**(digits - i - 1)
        else:
            data[num[i]] += 10**(digits - i - 1)

current = 9
result = 0
for i in sorted(data.values(), reverse=True):
    result += i * current
    current -= 1

print(result)
```

```python
n = int(input())

data = {}
for _ in range(n):
    num = input()

    digits = 0
    for i in num[::-1]:
        if i not in data:
            data[i] = 10**digits
        else:
            data[i] += 10**digits
        digits += 1

current = 9
result = 0
for i in sorted(data.values(), reverse=True):
    result += i * current
    current -= 1

print(result)
```

```python
from collections import defaultdict

n = int(input())

char_weights = defaultdict(int)

for _ in range(n):
    word = input()

    for pos, char in enumerate(reversed(word)):
        char_weights[char] += 10 ** pos

sorted_weights = sorted(char_weights.values(), reverse=True)
total = sum(weight * digit for weight, digit in zip(sorted_weights, range(9, 9 - len(sorted_weights), -1)))

print(total)
```
