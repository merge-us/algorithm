```python
n = int(input())

scores = [int(input()) for _ in range(n)]

next_score = scores.pop()

result = 0
while scores:
    current_score = scores.pop()

    if current_score < next_score:
        next_score = current_score
    else:
        next_score -= 1
        result += current_score - next_score

print(result)
```
