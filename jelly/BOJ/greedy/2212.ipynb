```python
n = int(input())
k = int(input())

data = list(map(int, input().split()))
data.sort()

distances = []

for i in range(n - 1):
    distances.append(data[i + 1] - data[i])

distances.sort()

for _ in range(k - 1):
    distances.pop()

print(sum(distances))
```

```python
n = int(input())
k = int(input())

data = list(map(int, input().split()))
data.sort()

distances = []

for i in range(n - 1):
    distances.append(data[i + 1] - data[i])

distances.sort(reverse=True)

print(sum(distances[k - 1:]))
```
