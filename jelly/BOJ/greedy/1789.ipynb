```python
s = int(input())

sum_result = 0

i = 1
while True:
    sum_result += i    

    if sum_result > s:
        break

    i += 1
    
print(i - 1)
```

```python
s = int(input())

left = 1
right = s
answer = 0

while left <= right:
    mid = (left + right) // 2
    total = mid * (mid + 1) // 2

    if total <= s:
        answer = mid
        left = mid + 1
    else:
        right = mid - 1

print(answer)
```
