```python
t = int(input())

for _ in range(t):
    n = int(input())
    prices = list(map(int, input().split()))

    max_revenue = 0
    
    current_max = prices.pop()
    while prices:
        current = prices.pop()
        if current < current_max:
            max_revenue += current_max - current
        else:
            current_max = current

    print(max_revenue)
```
