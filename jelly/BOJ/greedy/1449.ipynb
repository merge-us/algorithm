```python
n, l = map(int, input().split())

holes = list(map(int, input().split()))
holes.sort()

result = 1

current = holes[0]
for i in range(n):
    if holes[i] - current + 1 <= l:
        continue
    
    current = holes[i]
    result += 1

print(result)

    
    
```

```python
n, l = map(int, input().split())

holes = list(map(int, input().split()))
holes.sort()

result = 0
i = 0

while i < n:
    start = holes[i] - 0.5
    end = start + l
    result += 1

    while i < n and holes[i] + 0.5 <= end:
        i += 1

print(result) 
```
