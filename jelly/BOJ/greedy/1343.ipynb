```python
data = input()

board = data.split(".")
result = ""

if data[0] == ".":
    result += "."
    
for xs in board:
    len_xs = len(xs)
    current_result = ""
    
    current_result += "AAAA" * (len_xs // 4)
    current_result += "BB" * ((len_xs - len(current_result)) // 2)

    if len(current_result) != len_xs:
        print(-1)
        break

    result += current_result + "."
else:
    if data[-1] == ".":
        result += "."
    print(result)
        
```

```python
board = input().split(".")
result = []

for segment in board:
    length = len(segment)
    if length % 2 == 1:
        print(-1)
        break

    result.append("AAAA" * (length // 4) + "BB" * ((length % 4) // 2));
else:
    print(".".join(result))
        
```
