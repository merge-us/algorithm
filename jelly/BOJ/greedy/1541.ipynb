```python
data = input()

preprocessed = ""

c_operand = ""
c_operator = ""

for c in data:
    if c not in ["+", "-"]:
        c_operand += c
    else:
        preprocessed += str(int(c_operand))
        c_operand = ""
        if c == "-":
            c_operator = "-"
        else:
            if c_operator == "-":
                c_operator = "-"
            else:
                c_operator = "+"
                
        preprocessed += c_operator
        
preprocessed += str(int(c_operand))
print(eval(preprocessed))
```

```python
expr = input()

chunks = expr.split("-")
for i in range(len(chunks)):
    chunks[i] = eval(chunks[i])

result = chunks[0]
for i in range(1,len(chunks)):
    result -= chunks[i]

print(result)
    
```

```python
expression = input().split("-")

result = sum(map(int, expression[0].split("+")))
for part in expression[1:]:
    result -= sum(map(int, part.split("+")))
print(result)
    
```
