```python
from collections import Counter

name = list(input())

counter = Counter(name)

first = ""
mid = ""
last = ""

words = sorted(counter.keys())
for word in words:
    count = counter.get(word)
    if count % 2 == 1:
        if mid:
            print("I'm Sorry Hansoo")
            break
        mid += word
        count -= 1

    first += word * (count // 2)
else:
    for word in first[::-1]:
        last += word
    
    print(first + mid + last)
    
        
```
