```python
import sys

n, m = map(int, input().split())

a = [list(input()) for _ in range(n)]
b = [list(input()) for _ in range(n)]

count = 0
for i in range(n - 2):
    for j in range(m - 2):
        if a[i][j] != b[i][j]: 
            for k in range(3):
                for l in range(3):
                    a[i + k][j + l] = "0" if a[i + k][j + l] == "1" else "1"
            count += 1

check = True
for i in range(n):
    for j in range(m):
        if a[i][j] != b[i][j]:
            print(-1)
            sys.exit()
else:
    print(count)
        


```
