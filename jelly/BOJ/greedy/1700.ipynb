```python
n, k = map(int, input().split())
data = list(map(int, input().split()))

plug = []

count = 0
for i in range(k):
    if len(plug) < n:
        plug.append(data[i])
        continue
    
    if data[i] not in plug:
        target = data[i:i + k]
        for j in range(n):
            if plug[j] not in target:
                plug[j] = data[i]
                count += 1
                break

print(count)
                    
```

```python
n, k = map(int, input().split())
data = list(map(int, input().split()))

plug = []
count = 0

for i in range(k):
    if len(plug) < n:
        plug.append(data[i])
        continue
    
    if data[i] not in plug:
        target = 0
        latest = -1
        for j in range(n):
            next_use = k
            for l in range(i + 1, k):
                if plug[j] == data[l]:
                    next_use = l
                    break

            if next_use > latest:
                target = j
                latest = next_use
        plug[target] = data[i]
        count += 1

        
print(count)
                    
```
