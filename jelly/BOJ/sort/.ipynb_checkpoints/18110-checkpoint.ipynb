```python
import sys

n = int(input())

if n == 0:
    print(0)
    sys.exit()

data = [int(input()) for _ in range(n)]
data.sort()

excludes = round(n * 0.15)

sum_result = 0
for i in range(excludes, n - excludes):
    sum_result += data[i]

print(round(sum_result / (n - (2 * excludes))))
```

```python
import sys
import math

n = int(input())

if n == 0:
    print(0)
    sys.exit()

data = [int(input()) for _ in range(n)]
data.sort()

excludes = math.floor((n * 0.15) + 0.5)
excluded = data[excludes:n - excludes]

avg = sum(excluded) / len(excluded)
print(math.floor(avg + 0.5))
```
