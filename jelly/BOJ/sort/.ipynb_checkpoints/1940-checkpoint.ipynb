```python
n = int(input())
m = int(input())

data = list(map(int, input().split()))
data.sort()

i = 0
j = n - 1


count = 0
while i < j:

    while. data[i] + data[j] > m:
        j -= 1

    if data[i] + data[j] == m:
        count += 1

    i += 1

print(count)
```

```python
n = int(input())
m = int(input())

data = list(map(int, input().split()))
data.sort()

i = 0
j = n - 1


count = 0
while i < j:
    total = data[i] + data[j]

    if total == m:
        count += 1
        i += 1
        j -= 1
    elif total < m:
        i += 1
    else:
        j -= 1

print(count)
```
