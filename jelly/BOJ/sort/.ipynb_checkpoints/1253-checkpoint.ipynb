```python
from itertools import combinations
from bisect import bisect_left, bisect_right

n = int(input())
data = list(map(int, input().split()))
data.sort()

results = []
for num1, num2 in combinations(data, 2):
    target = num1 + num2
    count = bisect_right(data, target) - bisect_left(data, target)
    if count > 0 and target not in results:
        for _ in range(count):
            results.append(target)

print(len(results))
```

```python
n = int(input())
data = list(map(int, input().split()))
data.sort()

count = 0
for num in data:
    left = 0
    right = n - 1

    while left < right:
        target = data[left] + data[right]
        
        if num == target:
            count += 1
            break
        elif num > target:
            left += 1
        else:
            right -= 1

print(count)
```

```python
n = int(input())
data = list(map(int, input().split()))
data.sort()

count = 0
for i in range(n):
    target = data[i]
    left = 0
    right = n - 1

    while left < right:
        if left == i:
            left += 1
            continue
        if right == i:
            right -= 1
            continue
            
        total = data[left] + data[right]
        if total == target:
            count += 1
            break
        elif total < target:
            left += 1
        else:
            right -= 1

print(count)
```
