```python
from collections import Counter

n = int(input())

data = [input() for _ in range(n)]
data.sort()

counter = Counter(data)
print(counter.items())

max_item = None
max_count = 0
for item in counter:
    if counter[item] > max_count:
        max_item = item
        max_count = counter[item]

print(max_item)

```

```python
from collections import defaultdict

n = int(input())
books = [input() for _ in range(n)]

counter = defaultdict(int)
for title in books:
    counter[title] += 1

max_count = max(counter.values())
candidates = [title for title, count in counter.items() if count == max_count]

print(sorted(candidates)[0])
```
