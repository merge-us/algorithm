```python
from collections import Counter

n = int(input())

cards = [int(input()) for _ in range(n)]
counter = Counter(cards)

max_count = 0
max_cards = []
for card, count in counter.items():
    if count > max_count:
        max_count = count
        max_cards.append(card)

max_cards.sort()
print(max_cards[0])
```

```python
from collections import Counter

n = int(input())

cards = [int(input()) for _ in range(n)]
counter = Counter(cards)

max_cards = []
max_count = max(counter.values())
for card, count in counter.items():
    if count == max_count:
        max_cards.append(card)

max_cards.sort()

print(max_cards[0])
```

```python
from collections import Counter

n = int(input())

cards = [int(input()) for _ in range(n)]
counter = Counter(cards)

max_count = max(counter.values())
print(min(card for card, count in counter.items() if count == max_count))
```
